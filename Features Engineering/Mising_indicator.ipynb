```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import MissingIndicator,SimpleImputer
```

```python
df=pd.read_csv('train.csv',usecols=['Age','Fare','Survived'])
```

```python
df.head()
```

```python
df.info()
```

```python
x=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
x.head()
```

```python
y.head()
```

# Train test split

```python

x_train,x_test,y_train,y_test=train_test_split(
                             x,
                            y,
                            test_size=0.2,
                           random_state=42
                  )
```

```python
x_train.head()
```

# Simple Imputer class set 

```python
si=SimpleImputer()
x_train_tf=si.fit_transform(x_train);
x_test_tf=si.fit_transform(x_test);
```

# Logisticregration setup:

```python
from sklearn.linear_model import LogisticRegression
clf=LogisticRegression()
clf.fit(x_train_tf,y_train)
```

 ## Prediction using X_test dataset checking that the result of y_predict is same as y_test or not:

```python
y_pred=clf.predict(x_test_tf)
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```

## Now use Missing Indicator  check the same things the accuracy of the model improved or not:

```python
mi=MissingIndicator(); # mising indicator class
mi.fit(x_train);  # fit to mising indicator class
```

```python
x_train_missing=mi.transform(x_train) ##  create a new array that indicate that if age (as it has null) is null at an entry the this index of array  fill with true
x_test_missing=mi.transform(x_test)
```

```python
x_train_missing # see  here
```

```python
x_train['Age_na']=x_train_missing; # create new column at x train which shows that in which entry the valuse of age missing
x_train.head()
```

```python
x_test['Age_na']=x_test_missing; # create new column at x test which shows that in which entry the valuse of age missing
x_test.head()
```
