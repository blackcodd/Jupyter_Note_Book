```python
# Importing necessary libraries
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn import tree
import matplotlib.pyplot as plt

# Load the Iris dataset
iris = load_iris()
X = iris.data
y = iris.target
feature_names = iris.feature_names
class_names = iris.target_names

# Train a Decision Tree Classifier
clf = DecisionTreeClassifier(max_depth=3, random_state=0)
clf.fit(X, y)

# Option 1: Visualize using plot_tree (built-in, recommended)
plt.figure(figsize=(20,10))
tree.plot_tree(clf, 
               filled=True, 
               feature_names=feature_names, 
               class_names=class_names)
plt.show()
```

```python
import graphviz
from sklearn.tree import export_graphviz

# Export as DOT format
dot_data = export_graphviz(clf, 
                           out_file=None, 
                           feature_names=feature_names,
                           class_names=class_names,
                           filled=True, 
                           rounded=True, 
                           special_characters=True)

# Draw graph
graph = graphviz.Source(dot_data)
graph.render("iris_tree")  # Saves as PDF: iris_tree.pdf
graph  # Show in Jupyter
```

```python
# 1. Correct import
import dtreeviz

from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

# 2. Load data
iris = load_iris()
X, y = iris.data, iris.target
feature_names = iris.feature_names
class_names = list(iris.target_names)

# 3. Train the decision tree
clf = DecisionTreeClassifier( random_state=0)
clf.fit(X, y)

# 4. Create a dtreeviz visualization
viz_model = dtreeviz.model(
    clf,
    X_train=X, y_train=y,
    feature_names=feature_names,
    target_name="species",
    class_names=class_names,
)

# 5. Display in Jupyter Notebook
viz_model.view()

```
