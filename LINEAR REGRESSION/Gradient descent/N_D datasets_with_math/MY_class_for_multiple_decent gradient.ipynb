# NOW OUR WON CLass for GRADIENT DECENT:

```python
import numpy as np
class GDRegressor:
    def __init__(self,lerning_rate,epochs):
        self.intercept_=None # Randomly select
        self.coff_=None #Randomly select
        self.lr=lerning_rate
        self.epochs=epochs
    def fit(self,x,y):
        self.intercept_=0
        self.coff_=np.ones(x.shape[1])
        for i in range(self.epochs):
            y_pred= np.dot(x,self.coff_)+self.intercept_
            inter_derives=-2*np.mean(y-y_pred) 
            self.intercept_=self.intercept_-(self.lr*inter_derives)
            coef_dev=-2*np.dot((y-y_pred),x)/x.shape[0]
            self.coff_=self.coff_-(self.lr*coef_dev)
        print(self.intercept_)
        print(self.coff_)
    def predict(self,x):
        return np.dot(x,self.coff_)+self.intercept_
            
    

```

```python
gd=GDRegressor(0.5,1000)
```

```python
gd.fit(x_test,y_test)   # this is same as the code sklearm
```

```python
y_pred=gd.predict(x_test)
```
