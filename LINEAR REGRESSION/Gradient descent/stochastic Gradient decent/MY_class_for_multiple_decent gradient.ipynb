```python
import numpy as np
class SGDRegressor:
    def __init__(self,lerning_rate,epochs):
        self.intercept_=None # Randomly select
        self.coef_=None #Randomly select
        self.lr=lerning_rate
        self.epochs=epochs
    def fit(self,x,y):
        self.intercept_=0
        self.coef_=np.ones(x.shape[1])
        for i in range (self.epochs):
            for j in  range (x.shape[0]):
                idx=np.random.randint(0,x.shape[0])
                y_hat=np.dot(x[idx],self.coef_)+self.intercept_
                intercept_der=-2*(y[idx]-y_hat)
                self.intercept_=self.intercept_-(self.lr*intercept_der)
                coef_der=-2* np.dot(y[idx]-y_hat,x[idx])
                self.coef_=self.coef_-(self.lr * coef_der)
        print(self.intercept_)
        print(self.coef_)
    def predict(self,x):
        return np.dot(x,self.coef_)+self.intercept_
            
    

```

```python
sd=SGDRegressor(0.05,100)
```

```python
sd.fit(x_test,y_test)   # this is same as the code sklearm
```

```python
y_pred=sd.predict(x_test)
```
